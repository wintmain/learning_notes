# GPU制造商

在我们继续AMD/NVIDIA上的不同计算平台和模块之前，让我们先看看第一个GPU是如何产生的。

尽管术语GPU最早是由NVIDIA创造的，但IBM专业图形控制器（PGA）是1984年为PC发布的首批2D/3D视频卡之一。它使用一个内置的intel8088微处理器来处理所有与视频相关的任务，而不是用CPU来处理视频（例如绘制和着色填充多边形）。

由于5500美元左右的高价，以及在此期间与许多程序和非IBM系统的不兼容性，它无法在业界占据重要地位。但是，即便如此，PGA还是建立了一个标准化的硬件2D/3D加速模型，作为一个单独的板载处理器，这标志着GPU发展的一个重要步骤，即改进单独使用处理器进行图形计算的框架。

Silicon Graphics Inc.（SGI）于1989年发布了OpenGL；它作为一种跨平台的2D/3D应用程序编程接口（API）得到了广泛的应用和支持。因此，通过OpenCL的支持，SGI对现代图形硬件的设计发展也做出了重大贡献。

1993年，SGI发布了其RealityEngine图形处理板，其中包含不同的板和逻辑芯片。CPU处理上半部分的图形管道。在下半年，RealityEngine板接管了图形管道的各个后期阶段，每个阶段都有固定的数据流。SGI卡在90年代主要用于工作站。

上世纪90年代领先的3D图形硬件制造商，如3DFX、NVIDIA、ATI和Matrox，开始为消费者提供3D图形板。3DFX推出了 Voodoo 系列，NVIDIA发布了Riva TNT。ATI（此时已被AMD收购）推出了Rage，Matrox也发布了他们的Millennium系列卡。像《地震》和《末日》这样的游戏在这些平台上变得非常流行，将游戏行业推向了更高的高度，同时鼓励采用我们今天所知道的GPU。

## 第一代

3DFX伏都教在它的时代（1996年）非常流行。它被认为是第一个真正的三维游戏卡之一。它只提供三维加速，并依赖于一个单独的二维加速器。

其规格如下：
- 它通过PCI总线工作
- 它有一百万个晶体管
- 它有4MB的64位DRAM
- 它有一个50兆赫的核心时钟速度

Voodoo提供了纹理映射、z缓冲和光栅化，但它仍然依赖CPU进行顶点变换。

## 第二代

第一个GPU于1999年8月31日由NVIDIA发布，名为GeForce 256。它拥有2300万个晶体管、32MB 128位DRAM、120MHz核心时钟速度和四条64位渲染管道，每秒可处理数十亿次计算和1000万个多边形。

这一代卡首先使用加速图形端口（AGP）而不是PCI总线，它们在硬件中提供的新图形功能包括多纹理、凹凸贴图、光照贴图以及硬件几何变换和照明。

在这一代卡上，一旦程序员将图形数据发送到GPU的管道中，就不能对其进行修改。因此，这种管道被称为固定功能管道。这是一个缺点，因为它会导致不灵活的图形效果。这样的硬件不能利用OpenGL和DirectX API定义的最新标准，因此在这方面被落在了后面，尽管它们要快得多。

## 第三代

NVIDIA在2001年发布了geforce3，这使程序员能够在我们在上一节提到的以前不可编程的管道的部分上编程。他们现在可以通过着色器发送数据，并在管道中对其进行操作。着色器程序是用汇编语言编写的小内核。第一次，在管道的顶点处理阶段实现了有限的可编程性。NVIDIA的竞争对手ATI同时也为消费者提供了Radeon 8500卡。

GeForce 3 GPU拥有5700万个晶体管，64MB 128位DDR DRAM，核心时钟速度为200MHz。

RADEON8500GPU有6000万个晶体管，64MB的128位DDR DRAM，275MHz的核心时钟速度，还有128MB的变体。

## 第四代

第一个完全可编程的GPU在NVIDIA GeForce FX和ATI Radeon 9700发布时出现。

这些卡允许使用可编程顶点和像素着色器进行逐像素操作。它们允许有限的用户指定的输入输出数据操作映射，并作为单独的专用硬件，分配给像素着色器和顶点着色器处理。

2002年发布的ATI Radeon 9700拥有1.1亿个晶体管、128 MB 256位DDR DRAM和275 MHz核心时钟速度。

另外还有一款9500专业版，由于其8像素的管道和快速的性能，预算买家非常喜欢。一些电脑爱好者能够修改成9700非专业！

第一个GeForce FX（FX5200）于2003年发布，拥有4500万个晶体管、128MB 128位DDR DRAM和250MHz核心时钟速度。

随着2003年DirectX9的推出，GPU计算的时代刚刚到来！GPU硬件成为完全可编程的非图形访问（通用计算）。全浮点支持和高级纹理处理开始出现在卡片上。

## 第五代

根据摩尔定律，密集集成电路中的晶体管数量大约每两年翻一番。但从2003年到2004年，GPU硬件技术的发展速度比摩尔定律快得多。NVIDIA GeForce 6和ATI Radeon X800是2004年第一批使用PCI express总线的卡。早期的高级GPU语言，如Brook和Sh，都是为GPU编程而开发的。它们通过着色器程序中的真实条件、循环和动态流控制提供了更好的控制。在硬件方面，引入的一些新特性包括更高的精度（64位双支持）、多个渲染缓冲区、增加的GPU内存和纹理访问。

NVIDIA GeForce 6800 Ultra Extreme拥有2.22亿个晶体管，256 MB 256位GDDR3 DRAM，核心时钟速度为450 MHz。

ATI Radeon X800 XT Platinum拥有1.6亿个晶体管，256 MB 256位GDDR3 DRAM，核心时钟速度为520 MHz。

GPU可编程性的新趋势开始形成。GPU开始被视为一个伟大的工具，作为非图形应用程序的替代处理器。人们开始意识到，GPU可以作为大量可编程浮点马力和内存带宽，完全可以用于计算密集型计算！在通往第六代的路上！

# NVIDIA GPU上的计算

NVIDIA有三个著名的GPU平台，即GeForce、Quadro和Tesla，它们支持通用计算。GeForce系列的高端是消费级GPU，GPU计算爱好者通常对这些GPU感兴趣，以较低的预算范围运行GPU加速的应用程序（Titan系列除外）。从更高的预算角度来看，Quadro和Tesla产品线专门针对GPU加速计算应用。这类应用程序的许多功能仅在Quadro和Tesla GPU上可用。所有属于这三个平台的GPU在性能和特性上都有所不同。自诞生以来的几年中，它们已经过渡到不同的微体系结构。

## GeForce 架构

1999年推出（正如我们前面提到的），也是相当流行的游戏玩家的GPU阵容从NVIDIA，负担得起的GeForce GPU是什么使它有可能开始学习CUDA或测试您的GPU可编程应用程序。这是伟大的个人谁也碰巧是HPC爱好者，除了是热情的游戏玩家。更高的价格，GeForce Titan系列是专门满足这一要求的。

Titan V 属于NVIDIA Volta 架构。它有12GB的HBM2内存，1455MHz的提升时钟速度，640个tensor内核和5120个CUDA内核。

## Quadro 架构

Quadro产品系列主要集中于开发人员工作站上的计算机辅助设计（CAD）、计算机生成图像（CGI）和数字内容创建（DCC）的专业级图形渲染。第一个Quadro GPU于2004年推出。

使Quadro GPU在科学计算和机器学习方面领先于GeForce GPU的一个因素是高度优化的ECC内存和更大的浮点精度吞吐量。在所有三个GPU产品线中，Quadro系列是目前提供48 GB DDR6内存的产品！

Quadro RTX 8000属于NVIDIA Turing体系结构。它有一个48 GB的GDDR6内存和1770兆赫的ECC提升时钟速度，576个tensor内核和4608个CUDA内核。

 ## Tesla 架构

特斯拉GPU主要用于GPGPU和流处理，最初于2007年推出，基于以传奇电气工程师和科学家尼古拉·特斯拉命名的特斯拉微体系结构。尽管特斯拉C级GPU发布时带有双链路DVI输出，但通用规范没有任何输出端口。

特斯拉V100属于NVIDIA Volta架构。它有32 GB和16 GB的HBM2内存版本，具有1380 MHz的提升时钟速度、640个tensor内核和5120个CUDA内核。

由于它们纯粹是为了加速计算应用程序而制造的，因此当显示需求为零且计算任务是唯一优先级时，这样的gpu是最合适的。因此，特斯拉GPU特别专注于超级计算。

它们非常节能，产生的热量更少。这使得它们非常适合于行业级部署。

目前，特斯拉gpu可以极大地促进模拟和大规模计算（特别是浮点计算这是双精度可以是一个巨大的优势）。除了纯计算之外，还可以使用CUDA或ROCm为专业和科学领域的应用生成高端图像。

## GPUDirect

GPUDirect是一种NVIDIA技术，通过绕过CPU工作负载来提高gpu之间的传输速度，从而降低延迟。

直接内存访问（DMA）是一种允许计算机系统的硬件组件访问系统RAM而不必使用CPU的功能。

消息传递接口（MPI）是一种可移植的标准，用于通过进程及其资源调用可执行代码，以便在不同的并行计算体系结构上运行。通过调用程序通知进程。因此，它被称为消息传递。MPI标准是由学术界和工业界的研究人员共同设计的。

GPUDirect Peer-to-Peer支持高速DMA传输，以便在连接在同一系统主板上的两条PCIe总线上的两个GPU的内存之间复制数据。因此，这对在系统中有多个GPU的个人是有帮助的。

GPUDirect远程直接内存访问（RDMA）允许第三方设备（如固态驱动器（ssd）、网络接口卡（NICs）和InfiniBand适配器）跨远程系统直接访问多个GPU上的内存，从而降低MPI向GPU内存发送和接收消息的延迟。

GeForce gpu只支持GPUDirect点对点，而Tesla和Quadro也支持RDMA。

GPU Direct RDMA ASYNC允许GPU在没有任何CPU交互的情况下启动RDMA传输。

## SXM和NVLink

SXM代表符号扩展模式。NVLink，如NVIDIA所述，是一种高带宽和节能的互连，能够实现CPU到GPU和GPU到GPU的超高速通信。

NVIDIA为NVLink引入了一种新的SXM2夹层连接器，以解决外围组件Interconnect Express（PCIe）的问题，例如过载会导致高端GPU和多GPU系统的性能瓶颈。
titanv不支持nvlink2.0，但是它的CEO版（32gb版本）支持它。

我们在这里比较的三个GPU价格非常高：数千美元。在可负担的水平上，它可以被认为是可行的学术机构和行业，但不适合个人谁是小，独立的研究人员。

## NVIDIA CUDA

Compute Unified Device Architecture（CUDA）是NVIDIA在2007年为GPGPU编程发布的专有API。它允许您通过基于CUDA的C编程和预定义库来利用nvidiagpu的计算能力。随着NVIDIA关于API的详尽文档，CUDA从一开始就被学术界和工业界广泛采用。已经使用CUDA为GPGPU开发构建了几个实现。由于CUDA不是开源的，跨平台编程是不可能的。为了能够使用CUDA，拥有NVIDIA GPU是一个强制性要求。

# AMD-APU和GPU的计算

基于AMD的GPU可编程平台以异构系统架构（HSA）为中心。HSA是一组跨供应商的规范，允许在同一总线上集成cpu和gpu，共享内存和任务。HSA基金会继续由AMD和许多其他成员开发。AMD是HSA的创始成员之一。

HSA是从程序员的角度开发的，用于解决内存之间长时间数据传输的问题，特别是当涉及cpu和gpu时（这是在CUDA或OpenCL上编程时的标准）。

## APU中的GPU——APU设计的意义

APU设计的主要优点是CPU和GPU之间的距离大大缩短，因为两个处理器位于同一个单元上。HSA采用的这种模式允许CPU-GPU或APU内部通信的速度比传统方法快得多：CPU和GPU分别组装在主板上。所以，APU主板就是专门为此而设计的。

虽然APU是作为AMD产品销售的，但英特尔也按照同样的标准生产此类产品。

## AMD GPU–可编程平台

AMD有三个著名的GPU平台，即Radeon、Radeon Pro和Instinct。像NVIDIA一样，这些产品线经过多年的发展演变成了不同的Vega-GCN微体系结构。

### Radeon

Radeon实际上是由ATI在2000年推出的，2006年AMD接管ATI后，Radeon成为AMD的一个品牌。Radeon是AMD游戏玩家在消费者层面的首选。Radeon开发人员，也在消费者层面，现在也可以使用开源的AMD ROCm，这给他们开始学习GPU编程或测试GPGPU应用程序带来了额外的优势。我们之所以这样说，是因为Radeon VII的售价为699美元，1:4 FP64的性能为3.46 TFLOPS！

Radeon VII属于AMD Vega GCN 5.1体系结构。它有一个16gbhbm2内存，1750mhz提升时钟速度，3840个流处理器，60个计算单元，1tb/s内存带宽。

### Radeon Pro

Radeon Pro（FirePro的后继产品）系列与NVIDIA的Quadro卡竞争，并且主要专注于开发人员工作站上用于计算机辅助设计（CAD）、计算机生成图像（CGI）和数字内容创建（DCC）的专业级图形渲染。第一个Radeon Pro GPU于2016年4月作为Radeon Pro Duo发布。

使Radeon Pro GPU在科学计算和机器学习方面领先于Radeon GPU的一个因素是高带宽HBM2 ECC内存和更大的FP64功率。

Vega Frontier属于AMD Vega GCN 5.0架构。它拥有16gbhbm2内存、1600mhz的提升时钟速度、4096个流处理器、64个计算单元和484gb/s内存带宽。

### Radeon Instinct

Instinct GPU主要用于加速深度学习、人工神经网络和HPC/GPGPU应用，最初于2016年12月推出，基于GCN第三代和第四代微体系结构。

本能MI50属于AMD Vega 20 GCN 5.1架构。它拥有16gbhbm2内存、1746mhz提升时钟速度、3840个流处理器、60个计算单元和1tb/s内存带宽。它以6.7 TFLOPS（1:2）提供完整的FP64性能。

AMD的Instinct GPU是NVIDIA的特斯拉GPU的直接竞争对手，其制造目的也在于以不受限制的计算能力加速计算任务。

## AMD ROCm
