# 超越PC游戏的GPU计算世界

如果你是一个电脑游戏玩家，你必须非常熟悉图形卡的世界。根据它们的规格，您可能还熟悉它们如何影响您的游戏体验。让我们通过本章后面的章节深入探讨这个领域之外的内容。

# 什么是GPU？

顾名思义，GPU是一种电子电路，用作处理图形信息并输出到显示器上的处理器。这本书的范围是超越仅仅处理图形信息和步入`general purpose computing with GPUs (GPGPU)`的舞台。GPGPU的全部内容都是关于如何使用中央处理器（CPU）执行的，我们将在下一节中详细讨论。术语GPU和图形卡互换使用非常频繁。但事实上两者都有很大的不同。图形卡是作为GPU接口的平台。就像CPU位于主板插槽上一样，GPU位于显卡插槽上（与主板类似，我们可能认为它是迷你主板，但只是为了方便GPU及其冷却）。

计算呢？“computing”一词最明显地来源于“compute”一词。计算就是利用自己的硬件，借助自己的可编程进程来部署应用程序。可编程过程是由您定义的一组规则，它们随时可供您使用。当然，它们是基于您自己的算法，允许您根据手头的应用程序来满足自己的特定需求。

如果你从一个通用的角度来看计算，你会发现我们在上一段中提到的具体要求不仅仅局限于计算机科学。计算可以被推断为一种技术来计算任何可测量的实体，可以属于任何领域，无论是科学领域，甚至是艺术领域。既然我们已经分别描述了GPU和计算这两个术语，那么让我们继续介绍我们的主要主题：GPU计算（`GPU computing`）。

正如我们现在所理解的，GPU计算就是使用GPGPU和在GPU上执行的程序代码。当一个GPU程序员编写一个GPU程序时，主要的动机是移交一个特定的工作负载，这个工作负载在计算上比CPU要密集得多。

在代码中，CPU被指示将这些特定的操作交给GPU，然后由GPU进行计算。当这些计算完成后，GPU将所有这些信息发送回CPU，并将输出显示给您。由于计算结果要快很多倍，这样的工作也可以称为GPU加速计算（`GPU-accelerated computing`）。

# 传统的CPU计算——在GPU出现之前

在GPU出现之前，通用计算，正如我们所知，只有CPU才有可能实现，CPU是为消费者和高级计算爱好者制造的第一批主流处理器。

计算和图形处理都只由它们来处理。这意味着处理和处理输入计算以及在显示器上显示其相应的计算输出的任务都由CPU处理。

通用计算的历史可以追溯到20世纪50年代，在GPU到来并彻底改变这个概念之前。上世纪70年代见证了一个新时代的兴起，1971年英特尔发布了第一款商用CPU英特尔4004。第一台AMD CPU也在70年代推出，1975年推出AM2900。没有回头看，一个新的CPU制造周期开始生效，为每一代人带来了一系列新的微处理器。

虽然英特尔和AMD是CPU领域的热门竞争对手，但也有其他制造商，如Motorola（摩托罗拉）、IBM等。特别是Qualcomm（高通公司）和MediaTek（联发科），在移动通信行业占据主导地位。

因为这本书将是关于用Python进行GPU计算的，所以让我们简要回顾一下在Python开发任何GPU实现之前CPU计算是如何发展的。如果我们想了解CPU的计算能力，我们必须研究在GPU计算没有听说或部署之前，现代CPU是如何进化的。

自从第三代集成电路（IC）和微处理器问世以来，人们一直在思考，一个芯片可以投入多少能量才能获得最大的性能。在60年代早期，一块芯片曾经只包含几十个晶体管，但在70年代，这个数字上升到了几万个。在80年代，它变成了几十万个，而今天的芯片包含数十亿个晶体管。那么，你能在一个芯片上投入多少能量来获得最大的性能呢？

这就是CPU不断进化的原因。在此期间，英特尔和AMD都发明了新技术来改进CPU设计。由于在同一领域，他们于1981年签订了一项为期10年的协议，以实现相互间的技术交流。双核、酷睿2双核和许多其他技术变得流行起来。

但是，最终，人们认识到需要一种加速通用CPU计算的设备。那时GPGPU进入了竞技场，通用CPU的处理能力提高了10倍。

游戏行业是一个超过1000亿美元的产业。但早在20世纪50年代，电子游戏纯粹是出于学术目的。电子游戏是一种展示新发明技术能力的媒介。它们也是通过tic-tac-toe或chess测试AI应用程序的一个很好的应用程序。但这些平台的使用仍然局限于计算机实验室环境。

到了20世纪70年代，游戏领域开始发生变化。街机游戏变得非常流行。PC游戏格局在80年代形成，几乎家家户户都配备了可编程电脑，配备了Super Mario Bros（超级马里奥兄弟）、Donkey Kong（大金钢）、Prince of Persia（波斯王子）等流行游戏。

90年代出现了《末日》、《地震》等传奇游戏，从根本上改变了PC游戏的局面。许多PC爱好者和游戏玩家对了解PC硬件定制的好处产生了极大的兴趣。这类定制PC硬件的选项吸引了许多人的注意力，以实现流畅的游戏和最佳的视觉效果。

在这段时间里，游戏机市场也开始达到顶峰，一直持续到2000年代，品牌硬件作为一个单独的单元出货。后来，许多人开始好奇这些设备的规格，以了解它们的全部潜力，甚至在今天，当一个新的控制台上市时，这是一个非常常见的辩论，在新的控制台内的是否有GPU？

到2016年，游戏玩家超过20亿，其中一半生活在亚太地区。正如我们所看到的，游戏行业的兴起是众所周知的，而图形卡是充分利用游戏的必要条件，可以提供一些令人惊叹的视觉体验。

Integrated graphic（集成图形卡），就像在许多英特尔系统中看到的那样，跟不上游戏开发者或玩家的要求。因此，游戏行业开始大规模发展，GPU开始变得更便宜，从而形成了一个负担得起的GPGPU市场。以前，在PC游戏行业不那么流行的时候，它们的价格非常昂贵。计算机科学家们开始利用这一神奇的资源，于是开始了加速科学领域令人难以置信的冒险。

NVIDIA GeForce 3 Ti 200是有史以来第一款可编程GPU。

由于对PC游戏硬件的巨大需求以及GPU是令人惊叹的视觉效果的主要组成部分，图形卡成为每台用于视频游戏的PC的一个典型元素。

而且，技术爱好者社区中的许多人不仅仅是游戏玩家，其中许多人还包括相当多的程序员。就在那时，神奇的事情发生了，为GPU编程创建了一个新的社区。

从比较中我们可以看出，CUDA现在是GPU编程领域的领先API。CUDA提供的一些库对于线性代数、快速傅立叶变换、随机数生成和许多其他计算实现非常有用。
CUDA或OpenCL操作的基本思想包括三个步骤：
- 将数据（用于密集计算）从主存传输到GPU
- 使用CPU调用GPU内核对数据进行计算
- 计算结果后，它们将从GPU内存传输回主内存

自成立以来，CUDA得到了很大的学术支持，并得到了NVIDIA的支持。随着GPU一年比一年好，新技术和库也在不断发展，以最小的延迟最大限度地提高上一个进程的生产力。NVIDIA GPU根据其计算能力分为不同的组。如果我们考虑到在撰写本书时基于图灵体系结构的最新GPU的话，从1.0开始，最新的计算能力数字是7.5。

NVIDIA GPU的不同体系结构按时间顺序是Tesla, Fermi, Kepler, Maxwell, Pascal, Volta，最近的一个是Turing，它是多年来发布的。AMD GPU（以前称为ATI）的不同体系结构也是按时间顺序排列的，它们是R系列卡，从R100-R600、R670和R700开始，然后是Islands系列卡、Northern系列卡、Southern系列卡、Sea系列卡、Vancular系列卡和Artic GPU。自2017年以来，最新的架构被称为Vega。

2016年，AMD发布了一个名为GPUOpen的软件套件，可以用来创建通用的GPU计算应用程序。GPUOpen不是专有的，而是完全开源的。

# 人工智能的兴起和GPU的需求

CUDA为开发cuDNN等用于深层神经网络的库铺平了道路，作为机器学习的一种新方法和GPU人工智能编程的一部分，cuDNN在深层学习领域至关重要。cuDNN库可以在开发代码中安装和调用。

人工智能研究始于20世纪50年代，并导致了机器学习的产生。机器学习现在已经发展到深度学习，它使用神经网络实现在大数据集上训练人工智能算法。多年来，一些机器学习和深度学习库已经创建并正在积极开发中，如TensorFlow、Keras、Theano等。

作为这类机器学习库的一个子集，新的专门用于特定科学研究的库也被开发出来。Deepchem就是一个很好的例子，它是用机器学习库开发的，我们将在第12章，GPU加速使用Deepchem进行科学应用中进行研究。

创建这样的专用库使科学家更容易考虑开发有用的应用程序，而不是从头开始创建库。对于来自生物学等背景的科学家来说，这些工作更有意义，他们希望通过简化代码的理解，尽快部署这些应用程序。这样的应用也使得计算机科学家和生物信息学家之间的合作更加方便和容易。

那么，为什么GPU需要用于深度学习、机器学习或人工智能呢？在我们试图理解原因之前，重要的是要简单地知道什么是大数据。顾名思义，大数据涉及对海量数据进行数据挖掘和数据仓库的过程。以基因组学为例。在现代基因组学领域，一次简单的DNA测序产生的数据平均大小可达1TB。

计算数据不受任何特定领域的限制，不仅可以属于科学数据，也可以属于地理数据。想想世界地图可以生成的数据集。基于考古数据集的历史研究也可能相当庞大。以一幅价值连城的油画或艺术品的数字修复为例。图像数据可以是巨大的，但完美的使用图形处理器，它有巨大的内存量。最近的NVIDIA GPU有高达32 GB的视频内存！

GPU计算可以加速系统读取如此大数据的能力。当人工智能算法在数据集上进行训练时（这是深度学习的常规过程），引入GPU可以大大缩短训练时间，这就是为什么GPU在机器学习中的应用如此重要的原因。

今天的GPU有数千个内核，因此能够显著提高性能，比CPU快很多倍。出于同样的原因，许多机器学习开发人员现在为他们的库提供了单独的GPU版本。

由于人工智能加速是一种新的计算模式，人工智能时代刚刚到来，GPU是学生、专业人士和科学家学习或实施人工智能进行研究的最佳选择。

我们刚才提到的Volta架构是NVIDIA的一个以人工智能为核心的架构。NVIDIA的Volta微体系结构的GPU包含张量核，这是专为深度学习而设计的。这些张量核的引入使得深度学习应用程序的吞吐量和效率有了巨大的提高。这种新技术使Volta-GPU执行训练和推理任务的速度比Pascal架构的GPU快三倍。例如，NVIDIA Tesla V100是一款Volta GPU，包含640个张量核，其中数百个内核可以并行运行。Tesla也是NVIDIA早期微体系结构的名字，但它不再用于制造GPU。NVIDIA仍将其作为一个品牌，作为对科学界传奇人物尼古拉·特斯拉的致敬。

在人工智能竞赛中，谷歌凭借自己的一系列张量处理单元（TPU）遥遥领先。这些TPU专门构建为专用集成电路（ASIC），以加速基于神经网络的机器学习。由于这些TPU非常强大，液体冷却成为获得这些AI加速器最佳性能的必要步骤。至今已有三代TPU。最新的第三代TPU（3.0）可以执行高达100 petaflops（`计量单位，表示10的15次方，每秒千万亿次浮点运算`）的机器学习计算。

Python是一种编程语言，其语法非常容易掌握和理解，特别是对于来自计算机科学以外背景的计算分析人员。正是由于这个原因，它被全世界的研究界所采用。当我们还考虑到GPU强大的计算能力时，当结合Python语法的简单性时，一个双重优势是显而易见的。

## GPU的威力

GPU从仅仅用于图形应用的时候起已经走过了漫长的道路。几十年来，由于其相对于传统CPU的独特优势，其重要性在无限的应用领域继续受到关注。

如今的GPU拥有数千个内核，学术界和工业界都在不断开发GPU，以便在其专注于研究的应用程序中实现惊人的并行度。大量的CPU集群现在可以用几个或者，在某些情况下，仅仅一个GPU服务器来替代，以提供相同的生产力水平。

GPU已经使创造下一级超级计算机成为可能，以加速在不同领域的研究。大数据现在可以有效地计算与GPU并行使用。最新的NVIDIA的GPU架构是Turing，正如我们前面提到的。它的一些值得注意的事实和特点在下面几节中提到。

随着人工智能时代的到来，现在比以往任何时候都更需要GPU来处理具有各种深度学习技术的巨大数据集的训练。纵观迄今为止在这一领域取得的进展，我们显然走上了正确的道路。

### 光线追踪（Ray tracing）

简单地说，光线追踪，顾名思义，就是计算（追踪）光源路径的过程，以模拟光源在环境中不同物体上的行为。因此，光线追踪允许在使用图形引擎模拟三维环境时更有效地区分闪电和阴影场景。

NVIDIA最新推出的图灵GPU在其RTX系列卡中引入了这一功能，一旦启用，现代视频游戏和图形中的虚拟模拟将变得更加逼真。有许多较老的技术正在通过图形引擎使用，如HairWorks、PhysX和tressfx，这三种技术中的最后一种来自AMD。RTX-GPU有72个图灵RT核，可以提供高达每秒11g的实时光线跟踪性能。

## FP64在GPU计算中的意义

FP32是一种单精度浮点格式，需要分配32位内存。类似地，FP64是一种双精度浮点格式，需要分配64位内存。FP64允许在理论上以1:2 FP32进行高精度计算。

它涉及的计算位数是FP32的两倍，因此许多计算机科学家更喜欢具有最佳1:2 FP32性能的GPU。需要对物理环境进行建模和仿真的应用程序通常需要高性能的双精度精度计算。

因此，FP64 GPU就变得非常重要，并且绝对需要用于这些目的。但对于图像或统计数据的非计算处理，FP32就足够了。RTX卡系列不支持FP64。基于Volta架构的NVIDIA Titan V是支持FP64的最新NVIDIA GPU。

gpu可用于各种体育运动。在这里，我们分享一个足球比赛的例子。请注意，我们这里谈论的不是流行的EA FIFA电子游戏，而是真正的足球运动员！

该项目的目标是在不受控制的体育赛事环境下实时工作，如现场足球比赛。以高斯混合模型（GMM）为分割范式，通过分析足球相关的实时图像和视频，识别出由多种复杂颜色模式组成的足球运动员。包含的参考文献（图5）中提供了一幅实时演示分段使用的图像。

NVIDIA的CUDA平台加快了耗时的任务，后来又进行了重组和增强，大大加快了整个过程。在一个低成本的GPU上，该代码比CPU上的一个高度优化的C++版本快4到11倍，这两个版本在同一数据上被实时地以每秒64帧的速度进行处理。这项研究得出的一个重要结论是，应用程序的可扩展性与GPU上的内核数有关。

## 安全

本节讨论了以下研究论文：基于GPU的实时多摄像机视频分析系统（Real-time multi-camera video analytics system on GPU, P Guler, D Emeksiz, A Temizel, M Teke, T T Temizel (2013), Journal of Real-Time Image Processing, 11(3), 457-472. doi:10.1007/s11554-013-0337-2.）。

介绍了一种基于GPU的实时智能视频监控系统的并行实现。系统基于背景减法，由以下部分组成：
- 运动检测
- 摄像头破坏检测（摄像头可能会被移动、离焦或被遮挡）
- 废弃物体检测
- 目标跟踪算法

由于算法具有不同的特性，它们的GPU实现具有不同的加速率。当所有算法同时运行时，GPU中的并行化使系统的速度比CPU快21.88倍，从而可以对更多的摄像机进行实时分析。

## 农业

本节讨论了以下研究论文：《深度异常：结合背景减除和深度学习检测农田障碍和异常》，DeepAnomaly: Combining Background Subtraction and Deep Learning for Detecting Obstacles and Anomalies in an Agricultural Field, P Christiansen, L Nielsen, K Steen, R Jørgensen, H Karstoft (2016), Sensors, 16(11), 1904. doi:10.3390/s16111904 。

本文重点介绍了农民农田的视觉特征，在那里，障碍物，如人、动物和其他，很少发生。特别是，与空场（这是最常见的情况）相比，这些障碍物具有明显的外观。通过一种称为DeepAnomaly的算法，使用深度学习和异常检测来利用场的同质特征进行异常检测。

在人类探测器测试案例中，已经证明deepanomal比RCNN（一种以前存在的卷积神经网络，具有区域建议网络的实时目标检测）在更长的距离（45–90米）检测人类。RCNN在短距离（0-30米）也有类似的性能。然而，deepanomy的模型参数要少得多，每幅图像的处理时间（182 ms/25 ms）要快7.28倍。与大多数基于CNN的方法不同，它允许高精度、低计算时间，并且具有低内存占用，这使得它非常适合在嵌入式GPU上运行的实时系统。

自动无人机也被用于农业灌溉植物。它们的实现需要使用深度学习，因此在NVIDIA Jetson-TX的帮助下，GPU被用来加速这种复杂的计算。
