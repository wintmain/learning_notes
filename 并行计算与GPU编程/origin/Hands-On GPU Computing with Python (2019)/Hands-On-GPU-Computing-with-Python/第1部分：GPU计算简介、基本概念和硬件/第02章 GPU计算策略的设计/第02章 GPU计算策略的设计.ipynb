- 机箱

PC机箱、机箱或机柜是构建系统配置的地方。这个单位是什么将容纳你的主板及其互联组件，可以决定大小的情况下，你想要的，这取决于你的内部硬件的要求，将不得不在机箱容纳。市场上有不同形式的PC机箱，对应于标准ATX、Mini-ITX尺寸和microATX。

microATX主板始终与标准ATX机箱兼容。但是一个标准的ATX主板绝对不会放在microATX机箱里。因此，当购买PC机箱并考虑将来的升级时，如果您有microATX板，但以后需要升级需要标准ATX外形尺寸，那么选择标准ATX型号可能是您的首选。

在选择PC机箱时，如果您希望系统始终保持良好的通风和冷却（无论是在空闲状态还是满载状态），最佳冷却（空气或液体）选项也是一个重要的考虑因素。

- 电源设备（PSU）

这是一个非常重要的规范，您必须考虑，以确保您的所有系统都充分供电。许多人更喜欢使用在线电源计算器来估算定制PC所需的总功率。但是，除了了解功率要求外，对导轨、电压和电流的全面了解对于为系统选择最佳PSU起着至关重要的作用。我们将在本章的“自己动手”（DIY）桌面部分详细讨论它们。

- CPU

当您开发代码并调用GPU来交付高强度的计算任务时，CPU仍将负责获取计算结果并在显示器上报告它们。操作系统和应用程序实例都会因为CPU而运行，这就是为什么它应该擅长多任务处理。因此，为了充分利用GPU计算程序，选择一个同样兼容的CPU也是非常必要的。选择不匹配的CPU与高端GPU搭配会导致瓶颈问题，这会限制GPU的性能，从而使您无法充分利用它。

- RAM

取决于你的工作，一个基本的GPU计算设置为最低限度的学习目的，8-12GB的内存可以足够。但是，如果您希望执行一些重载实验，您可能希望用更多的RAM填充您的系统。如果您喜欢使用虚拟机（VM），则必须有足够的RAM来支持它们。我们将在下一节继续讨论更多内容。

- 硬盘驱动器（HDD）

在使用GPU计算系统时，大数据存储是首要考虑的问题，因为它涉及到大量的数据。当GPU处理大量数据用于计算或以数据集的形式用于机器学习时，高存储HDD满足了这一需求。

- 固态驱动器（SSD）

受欢迎主要是因为它的快速启动速度，固态硬盘也是首选的能力，推出应用程序相当快相比，硬盘，这是为GPU计算爱好者一个巨大的加分项。

- 监视器

一个好的监视器可以确保你能以更高的精确度来可视化你的三维结构。所以，这样的监视器可以很好地补充一个好的GPU。对于实验工作，最好至少在1080p显示器上工作。

# 构建第一台支持GPU的并行计算机–最低系统要求

在上一节中，我们简要讨论了所需配置的主要组件。由于此配置主要集中在GPU计算上，因此除了您的GPU之外，它还需要一组最低要求。对这些进行彻底的研究是非常必要的，不管你选择的是品牌预装配的桌面还是你自己组装的版本。

## 硬件可扩展性范围

最低系统要求将计算需要的范围设置为最低限度。这就是说，使用这样的配置，您至少应该能够在该硬件上执行您打算执行的特定类型的操作。为确保更好的稳定性，最好选择略高于最低系统要求的硬件配置。

根据你工作的性质，你可以把你的支出重点放在那一部分上。完成后，您可以继续相应地规划其余组件的费用。通过这种做法，你可以最大限度地超出你的预算，因为你是在你的电脑最重要的组成部分重点。

让我们用一个示例应用程序来看看它。假设您要开发一个应用程序，该应用程序涉及处理最小范围为1-2GB的数据。在这种情况下，您可能想考虑一个GPU至少有3 GB的视频内存。除此之外，如果您计划定期存储这样的输出，您还希望有足够的存储量。在这种情况下，您可以选择具有1-2 TB存储空间的硬盘驱动器。因此，除了GPU之外，一个5400 rpm的HDD至少有1 TB的存储空间就足够了。

如果我们考虑一个不同的例子，您可能希望选择SSD或7200rpm的HDD，如果更快的加载速度是您主要关心的问题，而不是存储时与您的GPU计算。

根据您的计算需要，我们建议您至少有8-12GB的RAM，以便顺利运行，此外还有一个当前或最新一代的CPU。这一步将允许您最大限度地提高计算效率。如果我们根据我们的计算需求小心地操作我们的系统组件，以最好的价格获得最好的性能将变得更加容易。

- 评估一个特定的配置及其价格也很重要，以确定您所支付的费用是否高于实际需要支付的费用

这两个步骤将帮助您在购买品牌GPU计算系统时降低成本。我们将在本章后面的单独一节中讨论支持品牌GPU的配置。

## 温度因素

当处理器忙于工作时，它们显然会在计算时不断产生热量。根据可变负载，温度也会受到影响。冷却背后的主要目标始终是保持这样的温度低，特别是当处理器在满负荷运行时。同时检查怠速温度始终是一种好的做法。

## 气流管理

管理任何冷却系统中的气流都需要一个风扇，该风扇可能连接到以下部件之一：
- 空气冷却时的散热器
- 散热器，在液体冷却的情况下
- 电脑机箱的前部、侧面或后部

考虑到前面三种情况，以下是配置风扇的不同方法：
- 推送配置包括通过附件将空气推入
- 拉动配置包括从附件中抽出空气
- 推拉配置包括通过附件的一侧将空气推入并从另一侧拉出空气
- 拉-推配置包括从附件的一侧抽出空气和从另一侧推入空气

这完全取决于你如何设置你的整个系统来决定配置和你使用的风扇类型。我们将在本章的“自己动手（DIY）”桌面部分阅读更多关于散热器、散热器的信息。

## 热糊

热糊是一种热化合物，在安装散热器或散热块之前涂在处理器表面。这种糊状物允许通过冷却装置有效地分配和散热。有些用户喜欢将糊状物均匀地铺在处理器表面，而有些用户则喜欢在中央形成豌豆形状，然后再将散热器或块安装在上面。

## 常规空气冷却

在我们深入研究液体冷却之前，有必要讨论常规空气冷却。传统的空气冷却通常包括一个散热器和一个风扇，风扇连接到处理器上，处理器的主板设计用于降低处理器运行期间的温度。当你购买一个CPU时，你也会在同一个包装内提供一个空气冷却器和一个散热器。当您购买图形卡时，GPU附带一个散热器和一个风扇作为一个单元安装在卡上，随时可以连接到主板。

## 库存冷却器

这两个CPU和GPU冷却器可以更具体地称为库存空气冷却器的技术基础上，其中长期库存涉及到他们的工厂发货状态。在CPU的情况下，你必须自己组装冷却器，但你不需要这样做你的GPU，因为它已经在一个单一的单位作为图形卡。因此，捆绑式冷却器被称为库存冷却器。

## 超频

超频是指在出厂时增加硬件部件的时钟速度等参数的过程，通常以提高其性能为目标。这些硬件组件可以是CPU、GPU，甚至是RAM模块。

关于超频组件的基本思想是在起步阶段改变其操作参数并检查其稳定性。当通过调谐电压、时钟速度和/或其他相关参数实现稳定状态时，我们在操作级别上实现该组件的超频状态。

## 那么，什么是定制/售后冷却器？

定制冷却器是指在PC硬件市场上可单独购买的冷却器。定制冷却器更普遍地被称为售后冷却器。在这些售后冷却器的帮助下，超频变得容易得多，因为当您手动更改硬件组件的操作参数时，操作温度会受到很大影响，而此时定制的冷却选项非常重要。因此，监控超频元件的温度可能更为关键。

通常不建议对库存冷却器进行超频。首选售后冷却器总是一个更好的选择，更适合，因为它是专为这样一个目的而设计的。在空气冷却和液体冷却，售后冷却器可用于CPU和图形卡。

空气冷却显然不仅仅局限于CPU或GPU；它还用于保持整个系统主板和机箱的环境温度。

## 液体冷却

液体冷却是在工作负载下使用液体作为处理器的循环冷却剂的过程。与其使用连接在风扇上的散热器，还不如使用冷却剂或冷却剂来达到同样的目的。冷却液在安装并连接到处理器的缸体中循环。一个或多个风扇，连接到散热器是这个循环回路的一部分。冷却回路可作为品牌的一体机（AIO）单元以及定制组件提供。

液体冷却可以大幅度降低温度，对于执行可能持续数天的长时间操作非常有用。由于液体冷却使温度非常低，这对处理器的健康和使用寿命非常有益。因此，许多研究人员、科学家和专业人士倾向于授权科学和专业应用程序，这些应用程序可能需要很多天才能完成。谷歌最新的张量处理单元（TPU）需要液体冷却以获得最佳性能，我们在上一章中提到了这一点。

这里要讨论的一点是，与液体冷却相比，空气冷却可能需要较少的维护，但维护后者是非常值得的。

## 冷却剂的比热容

比热容是将物质的温度提高1°C所需的热量。许多液体冷却爱好者在研究和寻找最佳冷却液时都会考虑到这个因素。

## Why is water the best liquid coolant?

在正常条件下，水具有最高的比热容（4.18 J/g/°C），也容易获得。由于它是如此容易获得，水是最好的形式的冷却剂提供给许多过度谁喜欢液体冷却。在水冷系统中，蒸馏水主要用作冷却剂。有些人还使用原水反渗透（RO）水。你可能会惊讶地知道，你可以暂时绕过过滤过程的家用反渗透水过滤器和收集原水反渗透（不含矿物质）为您的冷却剂的要求。

## PSU

在线使用电源计算器来计算您电脑的估计瓦数。之前，我们了解到，完全了解PSU中的轨道、电压和安培数对于为您的系统选择最佳PSU起着至关重要的作用。光看功率规格可能还不够。

导轨是PSU中的电线。如果一个PSU有三条提供12V电压的导轨，这意味着它有三条单独的导线，每条导线上都有12V电源。750 W PSU可能会被宣传，因此，即使它提供50 A的电流（电流），实际功率将如下所示：

$Power = Voltage X Current = 12V X 50 A = 600 W$

因此，将PSU宣传为750W而不是600W的背后原因将是在较低电压下提供了其他轨道。可能会有另一个轨道提供150瓦，比如说，5V x 30安。

所以，请详细检查这些规格，并知道最大真实瓦数当您购买您的PSU。这可能是非常关键的，因为高端GPU可能需要大量的电力。

它总是建议去至少一个青铜级的PSU，如果不是金的。选择一个模块化的电缆系统也将使您的工作更容易在电缆管理。

## 不间断电源（UPS）

选择一个好的UPS是一个明智的选择，尤其是当你计划在GPU计算系统上运行长时间的模拟或计算时。要估计所需的容量，可以使用以下公式：

最小伏安=电力负荷（瓦特）×1.6

因此，对于650 W的电力负载，您需要一个容量为650 W x 1.6=1040 VA的UPS。

备用电池的计算公式如下：

UPS备用=电池Ah X（电压/负载）X（1/功率因数）

平均功率因数为1.4。

或者，现在有许多UPS容量计算器在线，你饲料你的电脑规格，找出正确的UPS类型，你会需要。

## 冷却风扇类型

冷却风扇有不同的尺寸，从40、80、120、140、200到240毫米不等。

主要有两种类型的风扇：
- 高气流：我们已经熟悉这种风扇的设计，因为它们是非常传统和普遍的使用。气流风扇连接到处理器散热片和PC机柜壁上，在那里阻力较小，需要更多气流。
- 高静压：这些风扇的散热片设计的最好的方式来管理空气通过散热器在高静压。考虑到安装散热器时遇到的阻力，它们必须以不同的方式建造。

因此，对于空气冷却系统，高气流风扇将是首选。但在液体冷却系统中，高静压风扇是理想的选择。

## 瓶颈

这是一个概念，你必须记住，以确保您能够使用您的所有PC组件在其最大的潜力。当一个组件阻碍了整个系统按能力提供其最大潜力时，该组件被认为是在制造瓶颈。

CPU瓶颈的一个例子是使用旧一代CPU和当前一代GPU。无论GPU有多强大，您都无法有效地执行GPU计算操作，因为旧的CPU会阻碍它，因为它无法为GPU提供必要的操作。

类似地，GPU瓶颈的一个例子是使用具有当前一代CPU的旧一代GPU。

因此，在一起购买CPU和GPU以及升级其中任何一个CPU和GPU时，都要记住这一点。
