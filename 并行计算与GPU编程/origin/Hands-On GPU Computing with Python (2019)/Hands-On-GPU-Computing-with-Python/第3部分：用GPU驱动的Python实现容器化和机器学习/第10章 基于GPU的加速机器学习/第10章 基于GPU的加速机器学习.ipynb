# Python在人工智能中的意义——双重优势

在附录A中描述的场景中，GPU被证明是大数据管理的理想选择，而Python被证明是数据科学的理想选择。但事实上，反过来也是如此。因此，Python和GPU在数据科学和大数据中同样重要，这可以清楚地说是一种双重优势。

如果你仔细看一下附录A中引用的研究论文，你会发现Python和GPU在人工智能计算方面是相辅相成的。这一事实导致了在专门为GPU编写的Python现有框架中创建机器学习模块。在我们研究它们之前，我们必须了解双重优势的意义。

## 大数据管理的需求

为了说明一个大型数据集，让我们考虑一下癌症研究数据的例子。患这种致命疾病的病人每年只在增加。许多癌症研究仍在进行中，从机器学习的角度来看，仅在全球范围内，就一种癌症类型生成的数据量是巨大的。如果我们把它扩展到所有其他类型的癌症，数据量只会成倍增加；当我们从所有癌症类型转移到所有疾病类型时，数据量会进一步增加。所以现在，我们只关注大数据，因为被处理的数据将被处理、分析和存储在一个难以置信的大范围内。

这就是人工智能以其难以置信的处理能力拯救的地方，与Python协同工作，大数据处理变得更加容易。在人工智能中，机器学习算法在对大量数据集进行训练和有效推理时是最有效的。数据集的质量和数量越高，训练效果越好。只有这样，智能系统才能更好地发展。今天的GPU中提供的数千个内核使它们成为训练大型数据集的完美选择。

### 张量举例

您可能多次遇到这个术语，主要是因为流行的TensorFlow框架和TPU。在理解张量是什么之前，回顾一下我们高中时对标量和向量的理解是很重要的。标量只有大小但没有方向，而向量两者都有。

一个标量的例子就是温度。温度只有可以用摄氏度/华氏度/开尔文单位表示的量级，而没有任何方向属性。这样的标量可以称为秩为0的张量。

向量的一个例子就是速度。速度的大小和方向都可以用相对于时间的距离单位来表示。这样的向量可以称为秩1的张量。

秩为2的张量的一个例子是应力。应力可以表示为单位面积的有效力。但它可以在多个方向经历。

让我们用这个例子来说明这一点。下图展示了作用在立方体不同表面上的力。可见表面两侧的力由3个单位向量定义。在这3个表面上，1个力是通常的法向力，而另外2个力是侧向力。相反的力和扭矩在相对的6个方面确保机械平衡。因此，剩下的6个曲面不需要考虑。这种应力状态可分解为9（3x3）个分量，如下所示：

```python
fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
             'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']
plot.figure()
plot.imshow(train_images[0])
plot.colorbar()
plot.grid(False)
plot.show()
```

我们可以在这个图中看到，像素值在0到255之间：

在输入神经网络之前，图像被缩放到0到1的范围。我们对训练集和测试集都这样做，正如我们在这个代码块的前两行中看到的：

```python
train_images = train_images / 255.0
test_images = test_images / 255.0
plot.figure(figsize=(10,10))
for i in range(25):
    plot.subplot(5,5,i+1)
    plot.xticks([])
    plot.yticks([])
    plot.grid(False)
    plot.imshow(train_images[i], cmap=plot.cm.binary)
    plot.xlabel(class_names[train_labels[i]])
plot.show()
```

现在，我们从训练集中找到不同的服装相关项目。

现在我们开始构建模型，这是第一步：

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation=tf.nn.relu),
    keras.layers.Dense(10, activation=tf.nn.softmax)
])
```

现在我们编译模型，这是第二步：

```python
model.compile(optimizer='adam',
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy'])
```

前两步，构建（模型）和编译(模型.编译)模型，是非常重要的。特别是在TensorFlow中，不需要先建立和编译模型就可以训练模型，这样可以降低85%的精度

现在，我们通过一个单一的时代来测试训练。我们将其设置为epochs=1，以便快速查看流程在一个训练周期中的工作情况。请注意，训练是我们机器学习计划的第三步：

```python
model.fit(train_images, train_labels, epochs=1)
```

评估或测试准确性是第四步，正如我们在本节开头所看到的：

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print('Test accuracy:', test_acc)
```

根据物体的测试图像进行预测是我们的最后一步：

```python
predictions = model.predict(test_images)
print(predictions[0])
print(np.argmax(predictions[0]))
print(test_labels[0])

def plot_image(i, predictions_array, true_label, img):
    predictions_array, true_label, img = predictions_array[i],    true_label[i], img[i]
    plot.grid(False)
    plot.xticks([])
    plot.yticks([])
    plot.imshow(img, cmap=plot.cm.binary)
```

正确的预测用蓝色标记，不正确的预测用红色标记：

```python
predicted_label = np.argmax(predictions_array)
if predicted_label == true_label:
    color = 'blue'
else:
    color = 'red'
```

请注意，`predicted_label`由`predictions_array`定义：

```python
plot.xlabel("{} {:2.0f}% ({})".format(class_names[predicted_label],
                               100*np.max(predictions_array),
                               class_names[true_label]),
                               color=color)

def plot_value_array(i, predictions_array, true_label):
    predictions_array, true_label = predictions_array[i], true_label[i]
    plot.grid(False)
    plot.xticks([])
    plot.yticks([])
    thisplot = plot.bar(range(10), predictions_array, color="#777777")
    plot.ylim([0, 1])
    predicted_label = np.argmax(predictions_array)
    thisplot[predicted_label].set_color('red')
    thisplot[true_label].set_color('blue')
```

现在让我们看看预测：

```python
i = 0
plot.figure(figsize=(6,3))
plot.subplot(1,2,1)
plot_image(i, predictions, test_labels, test_images)
plot.subplot(1,2,2)
plot_value_array(i, predictions, test_labels)
plot.show()
```

在预测期间，真实描述符显示在括号中。在这里，我们看到它正确识别踝靴！

让我们看看下一个预测：

## 用PyTorch预测CIFAR-10

让我们来探索CIFAR-10数据集，但是使用PyTorch。我们将使用一个基于官方PyTorch文件的程序。

现在让我们进入我们为使用GPU设备而创建的经过大量修改的Pytork程序：

```python
import torch
import torchvision
import torchvision.transforms as transforms
```

变换用于规范化。另外，我们特别告诉PyTorch在通过`torch.device()`，而不是CPU：

```python
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

normalize = transforms.Compose(
  [transforms.ToTensor(),
   transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])
trainset = torchvision.datasets.CIFAR10(root='./pytorch', train=True, download=True, transform=normalize)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=4, shuffle=True, num_workers=2)
testset = torchvision.datasets.CIFAR10(root='./pytorch', train=False, download=True, transform=normalize)
testloader = torch.utils.data.DataLoader(testset, batch_size=4, shuffle=False, num_workers=2)
classes = ('plane', 'car', 'bird', 'cat',
         'deer', 'dog', 'frog', 'horse', 'ship', 'truck')
```

现在，我们导入Matplotlib.pyplot文件对于绘图：

```python
import matplotlib.pyplot as plot
import numpy as np

# functions to show an image
def imshow(img):
    img = img / 2 + 0.5 # unnormalize
    npimg = img.numpy()
    plot.imshow(np.transpose(npimg, (1, 2, 0)))
    plot.show()
```

我们得到一些随机训练的对象（船、鸟1、青蛙和鸟2）：

我们现在要预测四个这样的物体，它们属于10种CIFAR类型中的4种。我们将通过交叉验证训练模型和测试模型来实现这一点：

```python
# get some random training images
dataiter = iter(trainloader)
images, labels = dataiter.next()
# show images
imshow(torchvision.utils.make_grid(images))
# print labels
print(' '.join('%5s' % classes[labels[j]] for j in range(4)))
```

现在我们将导入`torch.nn`定义一个卷积神经网络。从`self.conv1`至`self.fc3`，我为使用GPU设备添加了`suffix .cuda()`：

```python
import torch.nn as nn
import torch.nn.functional as F
# Define a convolutional neural network to take 3-channel images
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.conv1 = nn.Conv2d(3, 6, 5).cuda()
        self.pool = nn.MaxPool2d(2, 2).cuda()
        self.conv2 = nn.Conv2d(6, 16, 5).cuda()
        self.fc1 = nn.Linear(16 * 5 * 5, 120).cuda()
        self.fc2 = nn.Linear(120, 84).cuda()
        self.fc3 = nn.Linear(84, 10).cuda()
    def forward(self, x):
        x = x.to(device)
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(-1, 16 * 5 * 5)
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

向前看，我们将本程序开头定义的设备传递给x。这确保我们在GPU上使用神经网络.

将参数和缓冲区转换为CUDA张量净额（设备）：

```python
net = Net()
net.to(device)

# Assuming that we are on a CUDA machine, this should print a CUDA device:
import torch.optim as optim

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.parameters(), lr=0.001, momentum=0.9)
```

现在我们定义训练周期，并在数据集上循环多次。输入和标签与GPU一起使用：

```python
for epoch in range(2): # loop over the dataset multiple times
    running_loss = 0.0
    for i, (inputs, labels) in enumerate(trainloader, 0):
        # get the inputs
        inputs, labels = inputs.to(device), labels.to(device)

        # zero the parameter gradients
        optimizer.zero_grad()

        # forward + backward + optimize
        outputs = net(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        # print statistics
        running_loss += loss.item()
        if i % 2000 == 1999: # print every 2000 mini-batches
            print('[%d, %5d] loss: %.3f' % (epoch + 1, i + 1, running_loss / 2000))
            running_loss = 0.0
```

打印统计数据后完成训练：

```python
print('Finished Training')
```

GroundTruth与我们的预测结果进行交叉验证：

```python
dataiter = iter(testloader)
images, labels = dataiter.next()

# print images
imshow(torchvision.utils.make_grid(images))
print('GroundTruth: ', ' '.join('%5s' % classes[labels[j]] for j in range(4)))
```

这四张图片分别是一只猫、一号船、二号船和一架飞机。我们可以辨认出两个不同的船图像：

我们将通过下一节了解它的预测，报告PyCharm控制台上的所有输出：

```python
outputs = net(images)

_, predicted = torch.max(outputs, 1)

print('Predicted: ', ' '.join('%5s' % classes[predicted[j]] for j in range(4)))
```

图像和标签也与GPU一起使用：

```python
correct = 0
total = 0
with torch.no_grad():
    for (images, labels) in testloader:
        images, labels = images.to(device), labels.to(device)
        outputs = net(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print('Accuracy of the network on the 10000 test images: %d %%' % (100 * correct / total))
```

在前面的代码中，我们报告了使用10000种CIFAR-10图像进行训练时网络的准确性：

```python
class_correct = list(0. for i in range(10))
class_total = list(0. for i in range(10))
with torch.no_grad():
    for (images, labels) in testloader:
        images, labels = images.to(device), labels.to(device)

        outputs = net(images)
        _, predicted = torch.max(outputs, 1)
        c = (predicted == labels).squeeze()
        for i in range(4):
            label = labels[i]
            class_correct[label] += c[i].item()
            class_total[label] += 1


for i in range(10):
    print('Accuracy of %5s : %2d %%' % (classes[i], 100 * class_correct[i] / class_total[i]))
```

最后，我们预测了前面代码中每种类型的CIFAR对象的准确度。
