# 为DeepChem解读科学概念

在你学习分子机器学习和DeepChem之前，我们必须学习一些在物理学、数学、生物学和最重要的化学中使用的基本术语：深层化学的基础是化学信息学。

化学信息学是计算机科学与化学相结合的交叉学科。它是一个不断发展的领域，涉及到开发计算方法和软件工具来分析和解释化学或生物化学数据。

在开始安装、配置和使用工具链之前，从这一点开始遵循的概念将帮助您更好地了解其用法。

## 原子

原子是具有基本化学性质的物质的最小单位。它由电子、质子和中子组成。质子和中子驻留在原子核上，而电子则围绕原子核旋转。例如，一个硅原子平均有14个中子、14个质子和14个电子。
