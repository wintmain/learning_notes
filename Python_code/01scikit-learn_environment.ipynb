```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
iris = load_iris()
print("iris 数据集的大小:",iris.data.shape)
print("目标数据集的大小:",iris.target.shape)
X_train,X_test,y_train,y_test = train_test_split(iris.data,iris.target,test_size=0.4,random_state=0)
print("生成的训练集的特征个数（数据个数）:",X_train.shape)
print("生成的训练集的标签个数（样本个数）:",y_train.shape)
print("生成的测试集的特征个数（数据个数）:",X_test.shape)
print("生成的测试集的标签个数（样本个数）:",y_test.shape)
print("iris数据集前五行数据:",iris.data[:5])
print("生成的训练集前五行数据:",X_train[:5])
print("生成的测试集前五行数据:",X_test[:5])
```
